# Will this number ever end? A binary example

Enter a decimal number to see if it can be precisely represented in binary or not. If it can't then Python will make an approximate representation which may lead to additional trailing numbers in calculations.

Run this Notebook from top to bottom, inputting the number you want to test in the second cell when the box asks for your input.

```python
from fractions import Fraction
```

```python
# Input test number
test = input("What number would you like to test? ")
```

```python
def to_binary(x):

    # Convert input into a numerator and denominator with the correct sign
    number = Fraction(str(x))
    sign = "-" if number < 0 else ""
    number = abs(number)

    # Divide find the divisor and remainder
    whole, remainder = divmod(number.numerator, number.denominator)
    result = sign + format(whole, "b")

    # Conditional if precisely divided
    if remainder == 0:
        return result

    result += "."
    seen = set()

    # Print under the repeated numbers are seen and then print result with elipses
    while remainder:
        if remainder in seen:
            return result + "..."

        seen.add(remainder)
        digit, remainder = divmod(remainder * 2, number.denominator)
        result += str(digit)

    return result
```

```python
to_binary(test)
```
