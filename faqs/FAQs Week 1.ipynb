### 1. Doubling down on quotation marks

When defining a string you can use either single or double quotation marks. This normally doesn't matter. In fact, the [PEP 8](https://peps.python.org/pep-0008/#string-quotes) style guide for Python says:

```
In Python, single-quoted strings and double-quoted strings are the same. 
This PEP does not make a recommendation for this. Pick a rule and stick to it. 
When a string contains single or double quote characters, however, use the other 
one to avoid backslashes in the string. It improves readability.

For triple-quoted strings, always use double quote characters to be consistent 
with the docstring convention in PEP 257.
```

The most important thing is to choose a convention and stick to it. With the exception of when you need to use many single or double quotes *within* your string

````python
# Normally quotation marks behave equally
print ('What is your name?')
print ("What is your name?")
````

````python
# When using a single quote as an apostrophe it needs to be encased in double quotes
print ("What's your name?")
````

````python
# Otherwise you get a syntax error as the string has been broken and Python doesn't know 
# how to intepret the rest of the string
print ('What's your name?')
````

````python
# When using double quotes to show it is  a quote
print ('Simon asked "What is your name?"')
````

````python
# Mixing double and single quotes in an outside of string markers means having to use a backslash
print ('Simon asked "What\'s your name?"')
````

I prefer using double quotes for strings but that is a personal preferce. Just pick what you like and stick to it but watch for mixed quotation mark usage!

### 2. Printing mixed types

In the Tutorial you were asked to print various different times. This means mixing integers and strings in a print statement. There are many ways to do this which all achieve the same result but there are a few nuances.

````python
hour = 10
minute = 25
````

````python
# Commas separate objects of different types so you don't have to cast integers to 
# strings but will add spaces by default
print (hour, ":", minute, "in the morning")
````

````python
# By altering the 'hidden' sep variable you can remove this space which will 
# remove all spaces
print (hour, ":", minute, "in the morning", sep="")

# You would have to manually add the space before the trailing text to fix this
print (hour, ":", minute, " in the morning", sep="")
````

````python
# Concatenating strings using + operators gives complete control over the spaces
print (str(hour) + ":" + str(minute) + " in the morning")
````

**Note:** You need to cast the integers to strings to concatenate

````python
# F-strings are the most flexible and readable alternative which use {} braces
# to encase variables of different types that will automatically be converted to strings
print (f"{hour}:{minute} in the morning")
````

The addition of the `f` before the quotation marks tells Python that this will be an `f-string`

F-strings are the most flexible option, are written an a readable format and avoid the challenges of casting while giving total control over spacing

### 3. Binary representations of floats

In some cases, calculations in Python can lead to results which look wrong to us due to differences in *precision*.

In one of the exercises, you were asked to raise a float to the power of 2 and 3. In some examples, this produces the expected result with no trailing numbers. In others, additional trailing numbers were added by Python even though we know the answer should not contain them. This is due to Python not having a concept of precise fractions. Python's `float` stores numbers in binary, so a decimal value is sometimes only an approximation to Python. When raising numbers to different powers, these approximations can magnify themselves.

These differences only appear sometimes and not others. This is becuase some floats can be expressed precisely in binary but others cannot.

````python
# This gives the expected precise value
x = 2.5
print (x**2, x**3)
````

2.5 can be expressed exactly in binary as 10.1 (2 + 0.5). It's square is 6.25 which is also exact in binary as 110.01 (4 + 2 + 0.25)

````python
# This gives additional trailing numbers
x = 2.2
# Mathematically we know this should be: 4.84, 10.648
print (x**2, x**3)
````

Here, 2.2 cannot be represented exactly in binary as 10.001100110011… (with 001 repeating forever). Similarly, the outputs of raising 2.2 to the powers of 2 and 3 cannot be represented exactly in binary.

By default, `float` in Python stores 53 binary digits (approx. 16 decimal places) so Python rounds the value of 2.2 to a value it can store. Then `print()` by default shows the shortest decimal that converts back to that same stored value which is 15 decimal places in each case.

This latter concept is referred to as `floating point precision`.
